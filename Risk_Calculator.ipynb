```python
risk_data = { "yaounde": 45, "douala": 52, "bafoussam": 28, "bertoua": 18, "garoua": 39, "limbe": 22, }
print("Cities Locations:", list(risk_data.keys()))
```

```python
location = input("Enter a city name: ").strip().lower()
print("You entered:", location)
```

```python
if location not in risk_data:
    print("City not found. Try: yaounde, douala, bafoussam...")
else:
    score = risk_data[location]
    print(f"Crime index for {location.title()}: {score}/100")
```

```python
if score > 40:
    status = "HIGH RISK"
else:
    status = "SAFE"
    
print(f"Location : {location.title()}")
print(f"Score    : {score}/100")
print(f"Status   : {status}")
```
